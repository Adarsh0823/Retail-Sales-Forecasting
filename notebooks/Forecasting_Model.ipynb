# Sales Forecasting Model

This notebook demonstrates step-by-step model training and evaluation for forecasting sales.

```python
# Import necessary libraries
import pandas as pd
from fbprophet import Prophet
import matplotlib.pyplot as plt

# Load data
data = pd.read_csv("../data/processed/sales_data_cleaned.csv")
data.rename(columns={'date': 'ds', 'sales': 'y'}, inplace=True)

# Display first few rows
data.head()
```

## Training the Prophet Model

```python
# Initialize and train the Prophet model
model = Prophet()
model.fit(data)
```

## Generating Future Forecasts

```python
# Create a dataframe for future dates
future = model.make_future_dataframe(periods=30)

# Generate forecast
forecast = model.predict(future)

# Display forecast
forecast[['ds', 'yhat', 'yhat_lower', 'yhat_upper']].tail()
```

## Visualizing the Forecast

```python
# Plot forecast
model.plot(forecast)
plt.title("Sales Forecast")
plt.show()
```

## Model Performance Metrics

```python
# Calculate Mean Absolute Error (MAE)
actual = data['y'][-30:].reset_index(drop=True)
predicted = forecast['yhat'][-30:].reset_index(drop=True)
mae = abs(actual - predicted).mean()
print(f"Mean Absolute Error (MAE): {mae}")
```
