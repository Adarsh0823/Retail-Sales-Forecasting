# Exploratory Data Analysis (EDA)

This notebook explores historical sales data to uncover trends, patterns, and correlations.

```python
# Import necessary libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from statsmodels.tsa.seasonal import seasonal_decompose

# Load data
data = pd.read_csv("../data/processed/sales_data_cleaned.csv")
data['date'] = pd.to_datetime(data['date'])
data.set_index('date', inplace=True)

# Display first few rows
data.head()
```

## Sales Trends Over Time

```python
# Plot overall sales trends
plt.figure(figsize=(12, 6))
data['sales'].plot(title="Sales Trends Over Time", xlabel="Date", ylabel="Sales")
plt.show()
```

## Seasonal Patterns

```python
# Decompose time series to observe seasonality
decompose = seasonal_decompose(data['sales'], model='additive', period=365)
decompose.plot()
plt.show()
```

## Correlations Between Features

```python
# Compute correlation matrix
correlation = data.corr()

# Heatmap of correlations
plt.figure(figsize=(8, 6))
sns.heatmap(correlation, annot=True, cmap="coolwarm", fmt='.2f')
plt.title("Feature Correlation Matrix")
plt.show()
```
